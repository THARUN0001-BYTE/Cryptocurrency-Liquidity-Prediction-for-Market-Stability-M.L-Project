# **Exploratory Data Analysis (EDA) Report**

**1. Introduction**

This document presents the Exploratory Data Analysis (EDA) for the project 'Cryptocurrency Liquidity Prediction for Market Stability'. The dataset includes cryptocurrency statistics such as price, market capitalization, and trading volume for two consecutive days (16/03/2022 and 17/03/2022). EDA provides insights into patterns, trends, and relationships within the dataset.

**2. Dataset Summary**

The dataset consists of multiple cryptocurrencies with features including:
- Price
- 24h Trading Volume
- Market Capitalization
- Derived Liquidity Ratio
- Moving Averages (2-day)
- Volatility (2-day)

The combined dataset merges data from two days into a single structure for analysis.

**3. Summary Statistics**

Basic descriptive statistics were computed for the key numeric columns:

Example (values for illustration):
- Price: min=0.05, max=42,000, mean=8,350
- 24h Volume: min=10,000, max=85,000,000, mean=12,400,000
- Market Cap: min=100,000, max=500,000,000, mean=80,000,000
- Liquidity Ratio: min=0.001, max=0.35, mean=0.05
- Volatility (2-day): ranged between 0.0 and 0.12

**4. Trends Over Time**

Analysis of the 2-day data revealed the following trends:
- Average cryptocurrency price remained relatively stable.
- Trading volume fluctuated more significantly than price.
- Market capitalization correlated strongly with trading volume.
- Liquidity ratios varied mostly due to volume changes.

**5. Correlation Analysis**

Correlation analysis showed:
- Market Cap and 24h Volume: Strong positive correlation.
- Price and Market Cap: Moderate positive correlation.
- Liquidity ratio depends more on 24h Volume than Market Cap.

***6. Distribution Analysis***

Distribution plots and histograms were generated:
- Market Cap distribution was right-skewed (few very large coins dominate).
- Liquidity ratio distribution showed that most coins fall in the medium range.
- Volatility distribution indicated that most coins had low volatility, with a few high-risk outliers.

**7. Key Insights**

- Liquidity ratios provide a better understanding of market stability than price alone.
- Larger market cap coins tend to have higher liquidity, but some smaller coins showed high ratios due to spikes in trading volume.
- Volatility remained low for most assets across the 2-day sample, indicating short-term stability.
