**Final Report: Cryptocurrency Liquidity Prediction for Market Stability**

**1. Executive Summary**

This final report summarizes the project 'Cryptocurrency Liquidity Prediction for Market Stability'. The goal was to develop machine learning models that predict cryptocurrency liquidity based on market data. The report consolidates Exploratory Data Analysis (EDA), High-Level and Low-Level Designs (HLD & LLD), the pipeline architecture, and model evaluation results.

**2. Project Objectives**

- Analyze cryptocurrency market data from two days (16/03/2022 and 17/03/2022).
- Preprocess and engineer features to reflect liquidity patterns.
- Train machine learning models (regression, classification, time-series forecasting).
- Evaluate models using standard metrics.
- Deploy the trained model locally for prediction via Flask/Streamlit.

**3. Dataset Description**

The dataset includes features such as:
- Price
- 24h Trading Volume
- Market Capitalization
- Liquidity Ratio (derived)
- Moving Averages (2-day)
- Volatility (2-day)

The combined dataset merges 16/03/2022 and 17/03/2022 into a single structure.

**4. Exploratory Data Analysis (EDA)**

- Prices ranged from $0.05 to $42,000, with a mean of $8,350.
- Trading volume varied from 10,000 to 85,000,000 (mean: 12,400,000).
- Market cap ranged from 100,000 to 500,000,000 (mean: 80,000,000).
- Liquidity ratios typically ranged between 0.001 and 0.35.
- Strong correlation observed between Market Cap and 24h Volume.
- Distribution analysis showed that most coins fell in medium liquidity bands.

**5. High-Level & Low-Level Design**

HLD (High-Level Design): Focuses on system architecture.
Raw Data → Preprocessing → Feature Engineering → Model Training → Evaluation → Deployment

LLD (Low-Level Design): Explains how each component works:
- Data Loader, Data Cleaner, Feature Generator
- Scaler, Model Trainer, Evaluator
- Deployment Wrapper exposing a /predict API.

HLD answers 'What the system does', while LLD answers 'How the system does it'.

**6. Pipeline Architecture**

The pipeline consists of:
1. Data Ingestion (Excel snapshots)
2. Preprocessing (cleaning, missing values)
3. Feature Engineering (ratios, volatility, moving averages)
4. Model Training (Linear Regression, Random Forest, Logistic Regression, Classifier)
5. Evaluation (RMSE, MAE, R², Accuracy, F1)
6. Deployment (Flask/Streamlit app)

Key performance values observed:
- Regression (RandomForestRegressor): RMSE=0.1230, MAE=0.0264, R²=0.8641
- Classification (RandomForestClassifier): Accuracy≈0.8878

**7. Model Training & Results**

Trained models:
- Linear Regression (baseline regression)
- Random Forest Regressor (best-performing regression)
- Logistic Regression (classification baseline)
- Random Forest Classifier (best-performing classifier)

Evaluation metrics:
- Regression: RMSE=0.1230, MAE=0.0264, R²=0.8641
- Classification: Accuracy=0.8878, strong F1 for high-liquidity classes.

**8. Insights & Limitations**

- Liquidity is influenced more by trading volume than market cap.
- Larger coins tend to have higher liquidity, but smaller coins can spike.
- Short-term data (2 days) limits forecasting power; more historical data needed.
- Time-series models (ARIMA, LSTM, Prophet) could be explored with extended data.

**9. Future Work**

- Integrate longer historical datasets.
- Explore deep learning (LSTM, Transformer models).
- Use hyperparameter tuning at scale.
- Build a live dashboard for real-time liquidity monitoring.

**10. Conclusion**

The project successfully built models to predict cryptocurrency liquidity, analyzed with strong accuracy and regression performance. EDA, HLD/LLD, and pipeline architecture were established, and the trained models were deployed locally for testing. This forms a foundation for more advanced, production-grade systems in the future.
