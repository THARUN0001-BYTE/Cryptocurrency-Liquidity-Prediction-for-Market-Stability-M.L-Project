# **Pipeline Architecture Document**

**1. Introduction**

This document describes the pipeline architecture for the 'Cryptocurrency Liquidity Prediction for Market Stability' project. It details data flow from raw input through preprocessing, feature engineering, model training, evaluation, and deployment.

**2. Pipeline Overview**

Pipeline steps (high-level):
1. Data Ingestion
2. Data Preprocessing & Cleaning
3. Feature Engineering
4. Model Training
5. Model Evaluation
6. Deployment

**3. Step-by-step Pipeline Details**

3.1 Data Ingestion:
- Source: Excel file with two snapshots (16/03/2022 and 17/03/2022)
- Action: Load and identify columns with _x and _y suffixes.

3.2 Data Preprocessing & Cleaning:
- Merge _x and _y datasets into a unified table.
- Convert numeric fields to proper dtypes.
- Handle missing values using median imputation.
- Standardize textual columns and parse dates.

3.3 Feature Engineering:
- Liquidity ratio: 24h_volume / mkt_cap
- Log transforms: log_price, log_volume, log_mktcap
- Rolling (2-day) features: ma2_price, ma2_volume
- Volatility: 2-day std of returns

3.4 Model Training:
- Models used: LinearRegression, RandomForestRegressor, LogisticRegression, RandomForestClassifier
- Training approach: For small data, train-on-full; for larger data, use train/test split and cross-validation.

3.5 Model Evaluation:
- Regression metrics: RMSE, MAE, R²
- Classification metrics: Accuracy, Precision, Recall, F1, Confusion Matrix

3.6 Deployment:
- Save trained models using joblib.
- Expose prediction endpoint using Flask (/predict) or build a Streamlit UI.

**4. Measurement & Values**

This section contains the key metric values observed during model evaluation (examples based on the current runs):
- Regression (RandomForestRegressor) → RMSE: 0.1230, MAE: 0.0264, R²: 0.8641
- Classification (RandomForestClassifier) → Accuracy: 0.8878 (approx), F1-scores vary per class

**5. Scalability & Monitoring**

- Monitor prediction drift & data drift.
- Retrain periodically as new data arrives (daily/weekly).
- For production, use containerization (Docker) and orchestration (Kubernetes) if needed.

**6. Conclusion**

The pipeline provides a clear and repeatable path from raw cryptocurrency data to liquidity predictions. With more historical data, time-series models and more robust validation can be added.
