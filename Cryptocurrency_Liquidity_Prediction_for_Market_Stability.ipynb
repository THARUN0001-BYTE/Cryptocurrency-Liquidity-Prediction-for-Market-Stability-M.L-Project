```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_squared_error, accuracy_score, classification_report
import pandas as pd
import numpy as np

# Load the dataframe
df = pd.read_excel("crypto_dataset.xlsx")

# Make sure date column exists
if 'date' not in df.columns and 'date_x' in df.columns:
    df['date'] = pd.to_datetime(df['date_x'])
elif 'date' not in df.columns and 'date_y' in df.columns:
    df['date'] = pd.to_datetime(df['date_y'])

df = df.sort_values(by='date')

#  Liquidity ratio ---
if '24h_volume_x' in df.columns and 'mkt_cap_x' in df.columns:
    df['liquidity_ratio_x'] = df['24h_volume_x'] / df['mkt_cap_x']

#  Moving averages (2-day since only 2 days available) ---
if 'price_x' in df.columns:
    df['ma2_price_x'] = df['price_x'].rolling(window=2, min_periods=1).mean()
if '24h_volume_x' in df.columns:
    df['ma2_volume_x'] = df['24h_volume_x'].rolling(window=2, min_periods=1).mean()

#  Volatility (2-day std of returns) ---
if 'price_x' in df.columns:
    df['return_x'] = df['price_x'].pct_change()
    df['volatility_2d_x'] = df['return_x'].rolling(window=2, min_periods=1).std()

#  Liquidity category labels ---
if 'liquidity_ratio_x' in df.columns:
    df['liq_label_x'] = pd.qcut(df['liquidity_ratio_x'], q=3, labels=['Low','Medium','High'], duplicates='drop')

# -------------------------
# Features & Targets
# -------------------------
features = ['price_x','24h_volume_x','mkt_cap_x','ma2_price_x','ma2_volume_x','volatility_2d_x']
X = df[features]

# Regression target: liquidity_ratio
y_reg = df['liquidity_ratio_x']

# Classification target: liquidity label
y_clf = df['liq_label_x']

# Drop rows with NaN values in features and corresponding targets
nan_rows = X.isnull().any(axis=1)
X = X[~nan_rows]
y_reg = y_reg[~nan_rows]
y_clf = y_clf[~nan_rows]


# Encode labels
le = LabelEncoder()
y_clf_enc = le.fit_transform(y_clf)

# -------------------------
# Train-Test Split
# -------------------------
X_train, X_test, y_reg_train, y_reg_test, y_clf_train, y_clf_test = train_test_split(
    X, y_reg, y_clf_enc, test_size=0.3, random_state=42
)

# Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# -------------------------
# Regression Models
# -------------------------
lr_reg = LinearRegression().fit(X_train_scaled, y_reg_train)
rf_reg = RandomForestRegressor(n_estimators=100, random_state=42).fit(X_train_scaled, y_reg_train)

# Evaluate Regression
for model, name in [(lr_reg, "Linear Regression"), (rf_reg, "Random Forest Regressor")]:
    y_pred = model.predict(X_test_scaled)
    mse = mean_squared_error(y_reg_test, y_pred)
    print(f"{name} MSE: {mse:.4f}")

# -------------------------
# Classification Models
# -------------------------
lr_clf = LogisticRegression(max_iter=500).fit(X_train_scaled, y_clf_train)
rf_clf = RandomForestClassifier(n_estimators=100, random_state=42).fit(X_train_scaled, y_clf_train)

# Evaluate Classification
for model, name in [(lr_clf, "Logistic Regression"), (rf_clf, "Random Forest Classifier")]:
    y_pred = model.predict(X_test_scaled)
    acc = accuracy_score(y_clf_test, y_pred)
    print(f"{name} Accuracy: {acc:.2f}")
    print(classification_report(y_clf_test, y_pred, target_names=le.classes_))
```

```python
# 1) Load file and build day_x / day_y frames (removes _x/_y suffixes)
import pandas as pd, numpy as np, os
path = ("crypto_dataset.xlsx")   # change if different
xls = pd.ExcelFile(path)
df = pd.read_excel(path, sheet_name=xls.sheet_names[0])

# collect columns that end with _x and _y
cols_x = [c for c in df.columns if c.endswith('_x')]
cols_y = [c for c in df.columns if c.endswith('_y')]

def build_day(df, cols, suffix, coin_candidates):
    tmp = df[cols].copy()
    tmp.columns = [c[:-2] for c in cols]            # drop suffix
    # attach coin column if available
    for cand in coin_candidates:
        if cand in df.columns:
            tmp['coin'] = df[cand]
            break
    # attach symbol if exists
    if 'symbol' in df.columns:
        tmp['symbol'] = df['symbol']
    # date detection: try exact date column with suffix or any date-like ending with suffix
    date_col = f"date{suffix}"
    if date_col in df.columns:
        tmp['date'] = pd.to_datetime(df[date_col], errors='coerce')
    else:
        found = next((c for c in df.columns if c.endswith(suffix) and 'date' in c.lower()), None)
        tmp['date'] = pd.to_datetime(df[found], errors='coerce') if found else pd.NaT
    return tmp

day_x = build_day(df, cols_x, '_x', ['coin','coin_x'])
day_y = build_day(df, cols_y, '_y', ['coin.1','coin_y','coin'])

print("day_x shape:", None if day_x is None else day_x.shape)
print("day_y shape:", None if day_y is None else day_y.shape)
# quick peek
display(day_x.head())
display(day_y.head())
```

```python
# Combine days, convert numeric columns, fill missing (median), basic feature engineering
```

```python
# 2) Combine both days -> df_all; convert numeric and fill missing with medians
df_all = pd.concat([day_x, day_y], ignore_index=True)

# Standardize column names expected later (ensure they exist)
# Typical numeric col names after suffix removal: price, 24h_volume, mkt_cap
expected_num = ['price','24h_volume','mkt_cap']
for c in expected_num:
    if c not in df_all.columns:
        raise KeyError(f"Expected numeric column '{c}' not found in combined data. Check column names.")

# convert to numeric safely
num_cols = expected_num.copy()
df_all[num_cols] = df_all[num_cols].apply(lambda s: pd.to_numeric(s, errors='coerce'))

# fill numeric NaNs with median (column-wise)
for c in num_cols:
    med = df_all[c].median()
    df_all[c].fillna(med if not np.isnan(med) else 0, inplace=True)

# ensure date column exists (fallback to assigning known dates if missing)
if 'date' not in df_all.columns or df_all['date'].isnull().all():
    # assign based on original grouping: first half -> 2022-03-16, second half -> 2022-03-17
    n = len(df_all)//2
    df_all['date'] = pd.to_datetime(np.where(df_all.index < n, '2022-03-16', '2022-03-17'))

df_all['date'] = pd.to_datetime(df_all['date'], errors='coerce')
df_all = df_all.sort_values(['coin','date']).reset_index(drop=True)

# quick check
print("Combined rows:", df_all.shape[0])
print(df_all[num_cols].head())

```

```python
# Liquidity features, log transforms, 2-day moving averages & volatility (per coin)
```

```python
# 3) Feature engineering: liquidity_ratio, logs, 2-day MA and 2-day volatility (per coin)
import numpy as np
import pandas as pd

# liquidity ratio
df_all['liquidity_ratio'] = df_all['24h_volume'] / (df_all['mkt_cap'].replace({0:np.nan}))

# log transforms
df_all['log_price'] = np.log1p(df_all['price'])
df_all['log_volume'] = np.log1p(df_all['24h_volume'])
df_all['log_mktcap'] = np.log1p(df_all['mkt_cap'])

# rolling features per coin (window=2 since we have two days)
df_all['ma2_price'] = df_all.groupby('coin')['price'].rolling(window=2, min_periods=1).mean().reset_index(level=0, drop=True)
df_all['ma2_volume'] = df_all.groupby('coin')['24h_volume'].rolling(window=2, min_periods=1).mean().reset_index(level=0, drop=True)

# returns & volatility (2-day rolling std of returns)
df_all['return'] = df_all.groupby('coin')['price'].pct_change(fill_method=None)
df_all['volatility_2d'] = df_all.groupby('coin')['return'].rolling(window=2, min_periods=1).std().reset_index(level=0, drop=True)

# fill any remaining numeric NaNs with median
all_num = ['price','24h_volume','mkt_cap','liquidity_ratio','log_price','log_volume','log_mktcap',
           'ma2_price','ma2_volume','volatility_2d']
for c in all_num:
    if c in df_all.columns:
        df_all[c].fillna(df_all[c].median(), inplace=True)

display(df_all[['coin','date','price','24h_volume','mkt_cap','liquidity_ratio','ma2_price','ma2_volume','volatility_2d']].head(10))
```

```python
# 4) Create categorical liquidity labels using quantiles and encode them
# Use rank + qcut to avoid errors with many equal values
df_all['liq_rank'] = df_all['liquidity_ratio'].rank(method='first')
df_all['liq_label'] = pd.qcut(df_all['liq_rank'], q=3, labels=['Low','Medium','High'])
# if qcut fails (e.g. identical ranks), fallback to simple binning
if df_all['liq_label'].isnull().any():
    df_all['liq_label'] = pd.cut(df_all['liquidity_ratio'], bins=3, labels=['Low','Medium','High'])

# encode labels for model training
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df_all['liq_label_enc'] = le.fit_transform(df_all['liq_label'].astype(str))

print("Label distribution:")
print(df_all['liq_label'].value_counts())
```

Easy to understand, but not always enough if classes are imbalanced.

**Precision**

Of all coins the model predicted as "High liquidity", how many are truly "High"?

Avoids false positives (wrongly labeling something as high when it isn’t).

**Recall**

Of all coins that are actually "High liquidity", how many did the model correctly identify?

Avoids false negatives (missing real high-liquidity coins).

**F1-score**

Balance between Precision and Recall.

Good when class sizes are imbalanced.

Confusion Matrix

Table showing counts of correct and incorrect predictions.

# **9. (" Model Testing & Validation")**: Test the model on unseen data and analyze predictions

**Since the dataset is very small (only 2 days), true “unseen data” is not available. But there some ways to handil Model Testing & Validation show you two ways to handle this:**

```python
# Train/Test Split (if you later have more days of data)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Split into train/test
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)

# Train on training set
rf_reg = RandomForestRegressor(n_estimators=100, random_state=42)
rf_reg.fit(X_train, y_train)

# Predict on unseen test set
y_pred = rf_reg.predict(X_test)

# Evaluate
mse = mean_squared_error(y_test, y_pred) # Get MSE
rmse = np.sqrt(mse) # Calculate RMSE manually
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Random Forest (Test Set) → RMSE: {rmse:.4f}, MAE: {mae:.4f}, R²: {r2:.4f}")
```

```python
# Cross-Validation (better when dataset is small)
```

```python
from sklearn.model_selection import cross_val_score, KFold
import numpy as np

cv = KFold(n_splits=3, shuffle=True, random_state=42)

rf_reg = RandomForestRegressor(n_estimators=100, random_state=42)

# RMSE (negative MSE, so take sqrt)
rmse_scores = np.sqrt(-cross_val_score(rf_reg, X_scaled, y, scoring='neg_mean_squared_error', cv=cv))
mae_scores = -cross_val_score(rf_reg, X_scaled, y, scoring='neg_mean_absolute_error', cv=cv)
r2_scores = cross_val_score(rf_reg, X_scaled, y, scoring='r2', cv=cv)

print("Cross-Validation Results (3-fold):")
print(f"RMSE: {rmse_scores.mean():.4f} ± {rmse_scores.std():.4f}")
print(f"MAE : {mae_scores.mean():.4f} ± {mae_scores.std():.4f}")
print(f"R²  : {r2_scores.mean():.4f} ± {r2_scores.std():.4f}")
```
