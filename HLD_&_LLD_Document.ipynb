# **High-Level Design (HLD) and Low-Level Design (LLD) Document**

**1. Introduction**

This document presents the High-Level Design (HLD) and Low-Level Design (LLD) for the project 'Cryptocurrency Liquidity Prediction for Market Stability'. The purpose is to provide both a system overview (HLD) and a detailed component-level breakdown (LLD).

**2. High-Level Design (HLD)**

The High-Level Design (HLD) focuses on the architecture and overall flow of the system. It provides an abstract view of the components and their interactions.

Key HLD Components:

1. Data Source: Cryptocurrency datasets for 16/03/2022 and 17/03/2022.
2. Data Preprocessing: Cleaning, handling missing values, formatting.
3. Feature Engineering: Adding liquidity ratio, moving averages, volatility.
4. Machine Learning Models:
   - Regression (Linear Regression, Random Forest Regressor)
   - Classification (Logistic Regression, Random Forest Classifier)
5. Evaluation: Using RMSE, MAE, R², Accuracy, Precision, Recall, F1.
6. Deployment: Flask/Streamlit app for local testing.

HLD Diagram (described):
Raw Data → Preprocessing → Feature Engineering → Model Training → Model Evaluation → Deployment

**3. Low-Level Design (LLD)**

The Low-Level Design (LLD) specifies the detailed steps, algorithms, and components used in the system.
LLD Breakdown:
- Data Loader: Reads Excel files and merges datasets.
- Data Cleaner: Removes duplicates, fills missing values.
- Feature Generator: Creates liquidity ratio, moving averages, volatility.
- Scaler: Standardizes numerical features with StandardScaler.
- Model Trainer:
   • Regression models (Linear Regression, Random Forest Regressor)
   • Classification models (Logistic Regression, Random Forest Classifier)
- Evaluator: Calculates RMSE, MAE, R², accuracy, precision, recall, F1.
- Deployment Wrapper: Flask API exposing /predict endpoint.

**4. HLD vs LLD Comparison**

HLD vs LLD Comparison:
- HLD gives the big picture of the architecture and main components.
- LLD explains the specific implementation details for each component.
- HLD answers 'What does the system do?'
- LLD answers 'How does the system do it?'

**5. Key Insights**

- HLD ensures all major modules and their interactions are clear.
- LLD ensures developers know exactly how to implement each module.
- Both together provide a complete design reference for the project.
